```python

# Week 2 - Python for Data Analysis
# Dataset: Sales Dataset

import pandas as pd

# --------------------------------------------------
# TASK 1: Load CSV and display basic information
# --------------------------------------------------

df = pd.read_csv("sales_dataset.csv")

# Clean column names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("FIRST 5 ROWS")
print(df.head())

print("\nDATASET SHAPE (Rows, Columns)")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES AND NON-NULL COUNTS")
df.info()

print("\nSUMMARY STATISTICS")
print(df.describe(include="all"))

# --------------------------------------------------
# TASK 2: Handle missing values and duplicates
# --------------------------------------------------

print("\nMISSING VALUES BEFORE CLEANING")
print(df.isnull().sum())

print("\nDUPLICATE ROWS BEFORE CLEANING")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing values
for column in df.columns:
    if df[column].isnull().any():

        if pd.api.types.is_numeric_dtype(df[column]):
            # Fill numerical missing values with median
            df[column] = df[column].fillna(df[column].median())

        else:
            # Fill text/date missing values with mode
            mode_values = df[column].mode(dropna=True)

            if not mode_values.empty:
                df[column] = df[column].fillna(mode_values[0])

print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())

print("\nDUPLICATE ROWS AFTER CLEANING")
print(df.duplicated().sum())

# Save cleaned dataset
df.to_csv("cleaned_sales_dataset.csv", index=False)

# --------------------------------------------------
# TASK 3: Group by category and calculate total revenue
# --------------------------------------------------

print("\nTOTAL REVENUE BY CATEGORY")

category_revenue = (
    df.groupby("category")["total_price"]
    .sum()
    .sort_values(ascending=False)
)

print(category_revenue)

# Save category revenue
category_revenue.to_csv("category_revenue.csv")

# --------------------------------------------------
# TASK 4: Sort by multiple columns
# --------------------------------------------------

print("\nSALES SORTED BY REGION AND TOTAL PRICE")

sorted_sales = df.sort_values(
    by=["region", "total_price"],
    ascending=[True, False]
)

print(sorted_sales.head(10))

# Save sorted data
sorted_sales.to_csv("sorted_sales.csv", index=False)

# --------------------------------------------------
# TASK 5: Correlation matrix for numerical columns
# --------------------------------------------------

print("\nCORRELATION MATRIX")

numeric_columns = df.select_dtypes(
    include="number"
).columns.tolist()

# Exclude order_id because it is an identifier,
# not a meaningful measurement
numeric_columns = [
    column for column in numeric_columns
    if column != "order_id"
]

correlation_matrix = df[numeric_columns].corr()

print(correlation_matrix)

# Save correlation matrix
correlation_matrix.to_csv("correlation_matrix.csv")

print("\nALL FIVE TASKS COMPLETED SUCCESSFULLY!")
```
